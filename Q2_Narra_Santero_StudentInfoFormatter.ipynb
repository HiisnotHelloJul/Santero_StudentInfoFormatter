```python
# Initial Inputs
first_name = input("Enter your first name: ")
last_name = input("Enter your last name: ")
section = input("Enter your section: ")
fav_sub = input("Enter your favorite subject: ")

name_code = [first_name, last_name]
charac_count = len(first_name) + len(last_name)

# Display Formatted Output
print("\n--FORMATTED OUTPUT--")
print(f"\nName: {first_name.strip().capitalize()} {last_name.strip().capitalize()}")
print(f"Name Code: {" ".join(name_code).strip().upper()} ")
print(f"Section: {section.strip().upper()}")
print(f"Favorite Subject: {fav_sub.strip().title()}")
print(f"Character Count: {charac_count}")
```
